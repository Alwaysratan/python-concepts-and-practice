## Key Takeaways

- Public members are accessible everywhere.
- Private members use `__` and help hide implementation details.
- Getters provide controlled read access.
- Setters provide controlled update access.
- Static variables belong to the class, not individual objects.

## Real-Life Analogy

Think of an ATM.

- You can withdraw money.
- You can deposit money.
- You cannot directly access or modify the bank's database.

The ATM provides controlled access to your account.

Similarly, in Python, we use methods (getters and setters) to control access to an object's data.
