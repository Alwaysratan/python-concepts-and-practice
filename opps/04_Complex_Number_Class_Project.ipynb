# question 1: class for complex numbers
"""
create a python class named `complexNumber` to relpresent complex numbers.
theory:
a comlex number is a number that comprises a real part and an imaganary part.
it is typically written in the form a+bi, where 'a' is the real part,and 'b'
is the imaganary part and 'i 'is the imagnary unit(√-1).
operations:

1.addition(+)

2.subtraction(-)

3.multiplication(*)

4.division(/)

5.comparision(==,!=)
"""


```python
class ComplexNumber:
    def __init__(self,real=0.0,imag=0.0):
        self.imag=imag
        self.real=real
    def __str__(self):
        if(self.real==0):
            s=f"{self.imag}i"
        elif(self.imag<0):
            s=f"{self.real}{self.imag}i"
        else:
            s=f"{self.real}+{self.imag}i"
        return s
    def __add__(self,other):
        realResult=0
        imagResult=0
        realResult=self.real+other.real
        imagResult=self.imag+other.imag
        ans=ComplexNumber(realResult,imagResult)
        return ans

    def __sub__(self,other):
        realResult=0
        imagResult=0
        realResult=self.real-other.real
        imagResult=self.imag-other.imag
        ans=ComplexNumber(realResult,imagResult)
        return ans
    def __mul__(self,other):
        realResult=0
        imagResult=0
        realResult=self.real*other.real - self.imag*other.imag
        imagResult=self.real*other.imag + other.real*self.imag
        ans=ComplexNumber(realResult,imagResult)
        return ans
    def __truediv__(self,other):
        realResult=0
        imagResult=0
        den=other.real**2+other.imag**2
        
        ans=self * ComplexNumber(other.real/den,-1*other.imag/den)
        return ans
        
        
    def conjugate(self):
        return ComplexNumber(self.real,-1*self.imag)
```

```python
#magic methods(dunder)
# they are special methods in python which are defined inside class.
# they login and end with doudle underscore(__)
# they have a special where in they get called themselves.
#     creation:-__init__
#     print:-__str__
#     +:-__add__
```

```python
c1 = ComplexNumber(3, 4)
c2 = ComplexNumber(1, 2)

# Addition
result = c1 + c2
assert result.real == 4
assert result.imag == 6

# Subtraction
result = c1 - c2
assert result.real == 2
assert result.imag == 2

# Multiplication
result = c1 * c2
assert result.real == -5
assert result.imag == 10

# Division
result = c1 / c2
assert abs(result.real - 2.2) < 1e-9
assert abs(result.imag + 0.4) < 1e-9

# Conjugate
result = c1.conjugate()
assert result.real == 3
assert result.imag == -4

# Additive identity
result = c1 + ComplexNumber()
assert result.real == 3
assert result.imag == 4

# Multiplication by zero
result = c1 * ComplexNumber()
assert result.real == 0
assert result.imag == 0

# Division by itself
result = c1 / c1
assert abs(result.real - 1.0) < 1e-9
assert abs(result.imag - 0.0) < 1e-9
```
