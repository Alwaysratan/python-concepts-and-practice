# Operator Overloading in Python

## What is Operator Overloading?

**Operator Overloading** is an OOP feature that allows us to change the behavior of built-in operators (`+`, `-`, `*`, `==`, etc.) for user-defined objects.

Python does this using **magic methods** (also called **dunder methods**) such as:

- `__add__()` → `+`
- `__sub__()` → `-`
- `__mul__()` → `*`
- `__truediv__()` → `/`
- `__eq__()` → `==`

---

# Why do we need Operator Overloading?

Suppose we create a class to represent a point.

Without operator overloading:

```python
point3 = point1.add(point2)
```

With operator overloading:

```python
point3 = point1 + point2
```

The second approach is much easier to read and write.

---

# Example 1: Overloading the + Operator

```python
class Number:
    def __init__(self, value):
        self.value = value

    # Overloading +
    def __add__(self, other):
        return Number(self.value + other.value)

    def display(self):
        print(self.value)


n1 = Number(10)
n2 = Number(20)

result = n1 + n2      # Calls __add__()

result.display()
```

### Output

```
30
```

### How it works

```text
n1 + n2
   │
   ▼
__add__(n1, n2)
   │
   ▼
10 + 20
   │
   ▼
30
```

# Example 4: Overloading the == Operator

```python
class Student:
    def __init__(self, marks):
        self.marks = marks

    def __eq__(self, other):
        return self.marks == other.marks


s1 = Student(90)
s2 = Student(90)
s3 = Student(75)

print(s1 == s2)
print(s1 == s3)
```

### Output

```
True
False
```

# Common Magic Methods

| Operator | Magic Method |
|----------|--------------|
| + | `__add__()` |
| - | `__sub__()` |
| * | `__mul__()` |
| / | `__truediv__()` |
| // | `__floordiv__()` |
| % | `__mod__()` |
| ** | `__pow__()` |
| == | `__eq__()` |
| != | `__ne__()` |
| > | `__gt__()` |
| < | `__lt__()` |
| >= | `__ge__()` |
| <= | `__le__()` |

---

# Important Points

- Operator overloading is achieved using **magic (dunder) methods**.
- It works only for **user-defined classes**.
- It makes code cleaner and more readable.
- The operator calls the corresponding magic method automatically.

---

# Interview Definition

**Operator Overloading** is the process of giving a new meaning to an existing operator when it is used with objects of a user-defined class. It is implemented using special methods (magic methods) like `__add__()`, `__sub__()`, and `__eq__()`.
