## Attributes

Attributes are variables that belong to an object.

They store information about that object.

```python
class Student:

    def __init__(self):
        self.name = "Ratan"
        self.age = 21
```

```python
student = Student()

print(student.name)
print(student.age)
```

## Methods

Methods are functions defined inside a class.

They describe the actions an object can perform.

```python
class Student:

    def study(self):
        print("Studying Python")

    def play(self):
        print("Playing Cricket")
```

```python
student = Student()

student.study()
student.play()
```

## Creating Multiple Objects

Each object is independent.

Changing one object does not affect another object unless they share class-level data.

```python
class Student:

    def __init__(self, name):
        self.name = name

    def display(self):
        print(self.name)
```

```python
student1 = Student("Ratan")
student2 = Student("Rahul")
student3 = Student("Aman")

student1.display()
student2.display()
student3.display()
```

## Key Takeaways

- A class is a blueprint for creating objects.
- An object is an instance of a class.
- `self` refers to the current object.
- Attributes store object-specific data.
- Methods define an object's behavior.
- Multiple objects can be created from the same class.
