# Inheritance in Python

## What is Inheritance?

Inheritance is an Object-Oriented Programming (OOP) concept that allows one class (child class) to acquire the properties and methods of another class (parent class).

It promotes:
- Code Reusability
- Easy Maintenance
- Better Organization
- Extensibility

---

## Syntax

```python
class Parent:
    pass

class Child(Parent):
    pass
```

---

## Basic Example

```python
class Animal:
    def sound(self):
        print("Animal makes a sound")

class Dog(Animal):
    pass

d = Dog()
d.sound()
```

### Output

```
Animal makes a sound
```

---

## Parent Class

A parent class (also called base class or superclass) is the class whose members are inherited.

```python
class Animal:
    def eat(self):
        print("Eating")
```

---

## Child Class

A child class (also called derived class or subclass) inherits the properties and methods of the parent class.

```python
class Dog(Animal):
    pass
```

---

## Why Use Inheritance?

- Avoid writing duplicate code.
- Reuse existing functionality.
- Makes programs easier to maintain.
- Allows extending existing classes.

---

# Accessing Parent Methods

```python
class Parent:
    def show(self):
        print("Parent Method")

class Child(Parent):
    pass

c = Child()
c.show()
```

### Output

```
Parent Method
```

---

# Constructor Inheritance

Child classes do **not** automatically call the parent's constructor.

Use `super().__init__()`.

```python
class Parent:
    def __init__(self):
        print("Parent Constructor")

class Child(Parent):
    def __init__(self):
        super().__init__()
        print("Child Constructor")

c = Child()
```

### Output

```
Parent Constructor
Child Constructor
```

---

# super()

`super()` is used to access the parent class methods and constructor.

## Calling Parent Constructor

```python
super().__init__()
```

## Calling Parent Method

```python
super().show()
```

---

## Example

```python
class Parent:
    def greet(self):
        print("Hello")

class Child(Parent):
    def greet(self):
        super().greet()
        print("Hi")

c = Child()
c.greet()
```

### Output

```
Hello
Hi
```

---

# Summary

- Inheritance allows one class to inherit another class.
- Parent class is also called Base Class or Superclass.
- Child class is also called Derived Class or Subclass.
- `super()` is used to access parent methods and constructors.
- Method overriding replaces the parent method.
- MRO defines the searching order for methods.
- Private variables are name-mangled.
- Protected members use `_`.
- Private members use `__`.
- `hasattr()`, `getattr()`, `setattr()` and `delattr()` are useful built-in functions for working with object attributes.
