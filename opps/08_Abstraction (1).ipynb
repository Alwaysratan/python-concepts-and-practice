```python
from abc import ABC,abstractmethod
```

```python
class User(ABC):

    @abstractmethod
    def login(self):
        pass
    def logout(self):
        print("logout")
    @abstractmethod
    def auth(self):
        pass 
```

```python
class Buyer(User):
    def login(self):
        if(self.auth):
            print("logging in User")
    def checkObject(self,link):
        print(link)
    def auth(self):
        return true
```

```python
class Buyer(User):
    def login(self):
        if(self.auth):
            print("logging in seller")
    def checkObject(self,link):
        print(link)
    def auth(self):
        print("otp")
        return true
```

```python
b1=Buyer()
```

so abstract classes are useful when we have a group of related objects that should
