```python

# What is Exception Handling?


# Exception Handling is used to handle runtime errors
# and prevent the program from crashing.

a = 10
b = 2

print(a / b)
```

```python

# Types of Errors


# 1. Syntax Error
# print("Hello"

# 2. Runtime Error
# print(10/0)

# 3. Logical Error
# Expected output is incorrect due to wrong logic.
```

```python

# try Statement


try:
    num = 10 / 2
    print(num)
except:
    print("An error occurred.")
```

```python

# except Statement


try:
    num = 10 / 0
    print(num)

except ZeroDivisionError:
    print("Cannot divide by zero.")
```

```python
try:
    number = int(input("Enter a number: "))
    print(number)

except ValueError:
    print("Please enter a valid integer.")
```

```python

# else Statement


try:
    num = int(input("Enter a number: "))
    print(num)

except ValueError:
    print("Invalid Input")

else:
    print("No Exception Occurred")
```

```python

# finally Statement


try:
    print(10 / 2)

except ZeroDivisionError:
    print("Cannot divide by zero.")

finally:
    print("Program Finished")
```

```python

# Multiple Exceptions


try:

    number = int(input("Enter a number: "))
    result = 100 / number

    print(result)

except ValueError:
    print("Invalid Number")

except ZeroDivisionError:
    print("Cannot Divide by Zero")
```

```python

# raise Statement


age = 16

if age < 18:
    raise ValueError("Age must be 18 or above.")

print("Eligible")
```

```python

# User-defined Exception


class InvalidAgeError(Exception):
    pass

age = 15

try:

    if age < 18:
        raise InvalidAgeError("Invalid Age")

except InvalidAgeError as e:
    print(e)
```

```python

# Common Built-in Exceptions


# ValueError
# TypeError
# ZeroDivisionError
# IndexError
# KeyError
# NameError
# FileNotFoundError
# AttributeError
# ImportError
# ModuleNotFoundError
```

```python
numbers = [10,20,30]

try:
    print(numbers[5])

except IndexError:
    print("Index Out of Range")
```

```python
student = {"name":"Ratan"}

try:
    print(student["age"])

except KeyError:
    print("Key Not Found")
```

```python
try:
    file = open("demo.txt","r")

except FileNotFoundError:
    print("File Does Not Exist")
```

```python
try:
    print(10 + "20")

except TypeError:
    print("Type Mismatch")
```

```python
# Build a Calculator with Exception Handling

try:

    num1 = float(input("Enter First Number: "))
    operator = input("Enter Operator (+,-,*,/): ")
    num2 = float(input("Enter Second Number: "))

    if operator == "+":
        print(num1 + num2)

    elif operator == "-":
        print(num1 - num2)

    elif operator == "*":
        print(num1 * num2)

    elif operator == "/":
        print(num1 / num2)

    else:
        print("Invalid Operator")

except ZeroDivisionError:
    print("Cannot Divide by Zero")

except ValueError:
    print("Please Enter Valid Numbers")

finally:
    print("Calculator Closed")
```
